# Metereologia CDMX

```python
import pandas as pd
```

```python
data = pd.read_csv('meteorologia_N6//meteorología_2025.csv')
```

```python
data = data.dropna()
```

```python
data.head()
```

```python
data['id_station'].groupby(data['id_station']).count().sort_values(ascending=False)
```

La estacion con la que vamos a trabajar es UAX, es la que mas valores tiene.

```python
df = data[(data['id_station'] == 'UAX')]
```

Hacemos una tabla donde cada columna sea el ambiente en filas ordenadas por el tiempo en el que se tomó.

```python
df_piv = df.pivot(index='date', columns='id_parameter', values='valor')
df_piv.head()
```

a) Probabilidad de humedad alta >= 70 (RH)

```python
p_hum_alta = (df_piv['RH'] >= 70).mean()
p_hum_alta
```

b) Probabilidad de viento tranquilo <= 1 (WSP)

```python
p_viento_tranq = (df_piv['WSP'] <= 1).mean()
p_viento_tranq
```

c) P(hum. alta | viento tranquilo)

```python
p_hum_dado_viento = (df_piv[df_piv['WSP'] <= 1]['RH'] >= 70).mean()
p_hum_dado_viento
```

d) P(viento tranq | hum. alta)

```python
p_viento_dado_hum = (df_piv[df_piv['RH'] >= 70]['WSP'] <= 1).mean()
p_viento_dado_hum
```
